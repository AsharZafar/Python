```python
tesla_data = yf.Ticker("TSLA")
```

```python
tesla_data.info
```

```python
Tesla_info=tesla_data.info
```

```python
import json

with open('Tesla_info','w') as json_file:
    json.dump(Tesla_info,json_file,indent=4)
```

```python
with open('/content/Tesla_info') as json_file:
    Tesla_data=json.load(json_file)
```
